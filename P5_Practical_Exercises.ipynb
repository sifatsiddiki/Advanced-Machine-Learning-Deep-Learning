##  **1. Importing the required libraries**

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

import numpy as np
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
%matplotlib inline
```

##  **2. Loading the data**
  First, we have to collect the IRIS dataset from Scikit-learn dataset pacakges using `load_iris()` function.

```python
# Loading IRIS dataset
iris = datasets.load_iris()
```

- Extracting data (X) and target (Y)  


```python
data = iris['data']
target = iris['target']

print ("Data (X): {}, Target (Y):{}".format(data.shape, target.shape))
print ("Format of Data (X): {} and Target (Y): {}".format(type(data), type(target)))
```

## **3. Normalise the data (X)**
  

```python
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)
```

## **4. Splitting data into train and test**

```python
X_train, X_test, y_train, y_test = train_test_split(data_scaled, target, test_size=0.2, random_state=2)
type(X_train), type(X_test), type(y_train), type(y_test)
```

## **5. Converting the numpy format data (X_train, X_test) to tensor**

```python
X_train = torch.FloatTensor(X_train)
X_test = torch.FloatTensor(X_test)
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
type(X_train), type(X_test), type(y_train), type(y_test)
```
